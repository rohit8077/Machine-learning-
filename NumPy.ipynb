```python
# Aggregating Function in NumPy
a = np.array([10,20,30,40])
print(np.max(a))
print(np.min(a))
print(np.sum(a))
print(np.size(a))
print(np.mean(a))
print(np.median(a))
print(np.cumsum(a))
print(np.cumprod(a))
```

```python
a = [10,50,90,80,70,30,102,60]
b = [1,5,9,8,7,3,12,6]
price = np.array(a)
quantity= np.array(b)

print(price ,"\n", quantity)
print(np.sum(price))
n= np.cumprod([price, quantity], axis = 0)
print(n[1].sum())
```

```python
#statistical Functions
import statistics as stats # for using the mode then we are import the statistics
Baked_food = [100,300,200,500,600,120,140]
a = np.array(Baked_food)
print(np.mean(Baked_food)) # Sum of all the value/number of Values
print(np.median(a)) # central value sfter sorting
print(stats.mode(a))
print(np.std(Baked_food))
print(np.var(Baked_food))
```
