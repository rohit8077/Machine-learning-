```python
data = {
    "Salary" : [25000, 30000, 35000, 35000,39000,45000, 50000, 51000, 67000,90000 , 100000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.shape
```

```python
# Line Plot:

plt.plot(df["Salary"], color="Purple", marker = "*", linestyle = ":" , linewidth = "2")
plt.grid()
plt.show()
```

```python
# Histogram

plt.hist(df ["Salary"], bins = 4 , color = "Green")
plt.grid()
plt.show()
```

```python
# Box plt

plt.boxplot(df["Salary"])
plt.show()
```

```python
df.loc[20] = [0]
```

```python
df["Age"] = [18, 20, 22, 25, 30, 35, 48, 40, 32, 45, 40 ]
```

```python
df.head()
```

```python
# Scatter Plot :

plt.scatter(df["Salary"], df["Age"], color = "Orange")
plt.show()
```

```python
#Line plot:
plt.plot(df["Age"], df["Salary"], color = "Orange", marker="*" , linestyle = ":", linewidth= "2")
plt.grid()
plt.show()
```

```python
sort_age = df.sort_values("Age")
```

```python
#Line plot

plt.plot(df["Age"],df["Salary"], color = "Green" , marker = "*", linewidth = "2")
plt.grid()
plt.show()
```

```python
#Bar Chart

plt.bar(df["Age"], df["Salary"], color="Green", align = "center")
plt.show()
```

```python
# bivariate : Numerical - categorical
```

```python
it_sal = df[df["dept"]=="IT"] ["Salary"]
it_sal
```

```python
hr_sal = df[df["dept"]=="HR"] ["Salary"]
hr_sal
```

```python
Finance_sal = df[df["dept"]=="Finance"] ["Salary"]
Finance_sal
```

```python
#Boxplot

plt.boxplot ([hr_sal, it_sal, Finance_sal],labels=["HR", "IT", "Finance"])
plt.show()
```

```python
#pie chart

salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels = salary_by_dept.index, autopct = "%1.2f",shadow= True , explode =[0.1,0.1,0.1])
plt.axis("equal")
plt.show()
```

```python
# bar plot

hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
Finance_mean = sum(Finance_sal)/len(Finance_sal)
```

```python
plt.bar(["HR", "IT", "Finance"],[it_mean, hr_mean, Finance_mean], color = ["green", "Red", "Yellow"])
plt.grid()
plt.show()
```

```python
#Multivarite Analysis: 3 numerical columns
```

```python
df.head()
```

```python
df["experience"] = [1,2,3, 5.6,6,3,7,12,3,9,8]
```

```python
df.head()
```

```python
plt.scatter(df["Age"], df["Salary"], s=df['experience']*20,color="Blue", edgecolors = "black")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# 2 numerical and 1 categorical column:
```

```python
plt.scatter(df["Age"],df["Salary"], c = df["dept"].map({"HR":"Yellow","IT":"Blue","Finance":"Orange"}))
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary vs Dept")
plt.legend()
plt.show()
```

```python
color = {"HR":"Yellow","IT":"Blue","Finance":"Orange"}

for dept , color in color.items():
   df_dept = df[df["dept"]== dept]
   plt.scatter(df_dept["Age"], df_dept["Salary"], color = color, label = dept)
legend = plt.legend()
```

```python
#object Oriented API
```

```python
fig, axs=plt.subplots(1, 3,figsize = (10,5))
#Line plt
axs[0].plot(sort_age["Age"], df["Salary"], color = 'red', marker = "*", linewidth = "2", markersize = 2)
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#histogram

axs[1].hist(df["Salary"], bins =5, color= "Green")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

#Boxplot

axs[2].boxplot(df["Salary"])
axs[2].set_title("Boxplot")
axs[2].set_xlabel("Salary")
plt.savefig("multipleplots.png")
plt.show()
```

```python
data= {
    'Year':[2020,2021,2022,2023],
    'Sales':[100,200,250,300],
    'Profit':[20,30,40,50],
    'Expenses':[80,120,160,200]
}
```

```python
d2f = pd.DataFrame(data)
```

```python
plt.plot(d2f["Year"],d2f["Sales"], label ="Sales")
plt.plot(d2f["Year"],d2f["Profit"], label ="Profit")
plt.plot(d2f["Year"],d2f["Expenses"], label ="Expenses")

plt.title ("Financial Analysis")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
plt.show()
```
