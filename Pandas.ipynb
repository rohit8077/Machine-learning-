# **Mearging and Concat**

```python
import pandas as pd
as1 = pd.DataFrame({"A":[1,2,3,4],"B":[5,6,7,9]})
as2 = pd.DataFrame({"A":[1,2,3,5],"C":[15,16,17,18]})

pd.merge(as1,as2, left_index = True, right_index = True , suffixes =("name", "id"))
```

```python
sr1 = pd.Series([1,2,3,4])
sr2 = pd.Series([5,6,7,8])

pd.concat([sr1,sr2])
```

```python
d1= pd.DataFrame({"A":[1,2,3,4],"B":[5,6,7,8]})
d2= pd.DataFrame({"A":[1,2,3],"B":[11,12,14]})
pd.concat([d1,d2], axis=1, keys=["d1", "d2"])
```

# **JOIN**

```python
import pandas as pd

z1 = pd.DataFrame({"A":[1,2,3,4],"B":[11,12,13,14]}, index=["a","b","c","d"])
z2= pd.DataFrame({"C":[10,20,30],"D":[21,22,23]},index=["a","b","c"])
z2.join(z1)
```

# **Melt()**

```python
import pandas as pd
var = pd.DataFrame({"Days":[1,2,3,4,5,6], "Eng":[10,11,12,13,14,15],"Maths":[17,18,19,20,21,22]})
var
```

# **pivot**()

```python
import pandas as pd
var = pd.DataFrame({"days":[1,2,3,4,5,6],
                    "std_name":["a","b","c","a","b","c"],"eng":[10,11,12,13,14,15],"maths":[17,18,19,20,21,22]})
var
```

```python
var.pivot(index="days", columns="std_name",values="eng")
```

```python
var.pivot_table(index="std_name",columns="days",aggfunc="sum", margins=True)
```
